# Training an NMD-efficiency predictor

This notebook describes the training and analysis of machine-learning models for predicting nonsense-mediated mRNA decay (NMD) efficiency. The implementation builds directly upon the original NMD-efficiency study (<https://github.com/hjkng/nmdeff>) and extends it by incorporating variant-level annotations generated with the NMD-Scanner.

All model training and evaluation is performed using the official benchmark datasets provided by the original NMD-efficiency study, namely the TCGA and GTEx datasets. The original training script (`train.ipynb`) from the NMD-efficiency repository (https://github.com/hjkng/nmdeff) was used as a reference implementation and starting point.

To assess the predictive value of NMD-Scanner annotations, variants from the original benchmark datasets were extracted, annotated using the NMD-Scanner, and merged with the corresponding benchmark datasets to obtain experimentally measured NMD_efficiency values. Importantly, the target variable remains unchanged, and only the feature representation is modified.

Throughout this notebook, multiple regression models and feature representations are evaluated, and the same analysis workflow is applied to all training scenarios to ensure comparability.


## Overview of the training and analysis workflow

For each model training scenario described below, the same analysis workflow is applied:
1. Training of a regression model to predict *NMD_efficiency*
2. Assessment of the target distribution in the respective dataset
3. Analysis of feature–target correlations using Pearson correlation coefficients
4. Evaluation of model-derived feature importance values

This consistent workflow enables a direct comparison of feature relevance and model behavior across different datasets and annotation sources.

## Training scenarios

The following training scenarios are investigated in this notebook:
1. Reproduction of the original NMD-efficiency prediction model using the TCGA benchmark dataset
2. Training using NMD-Scanner annotations on the TCGA benchmark dataset
3. Training using NMD-Scanner annotations on the GTEx benchmark dataset
4. Comparison of multiple regression models and hyperparameter optimization to select and train the final NMD-efficiency prediction model

## 1. Train NMD efficiency Predictor upon RandomForestRegressor

```python
import pickle
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from scipy.stats import pearsonr, spearmanr
import joblib
import os
```

```python
# load dataset
df= pd.read_csv('~/NMD/nmd-variant-effect-prediction/resources/TCGA_benchmark/tcga_dataset.csv')
df.shape
```

```python
'''
Description of columns
        downstream_exon_count (as Downstream exon count): The number of exons downstream of the PTC
        last_exon (as Last exon): 1 if the PTC is located on the last exon; 0 otherwise
        PTC_to_start_codon (as Dist PTC to start codon): The distance between the PTC to start codon
        dist_to_stop_codon (as Dist PTC to normal stop codon): The distance between the PTC and normal stop codon
        PTC_exon_length (as PTC-containing exon length): The length of the PTC-containing exon
        PTC_to_intron (as Dist PTC to downstream EJ): The distance between the PTC and downstream exon junction
        upstream_exon_count (as Upstream exon count): The number of exons upstream of the PTC
        mRNA_half_life (as mRNA half-life): The half-life of mRNA
        50nt_to_last_EJ (as Last 50nt penultimate exon): 1 if the PTC is located on the last exon; 0 otherwise
        LOEUF (as LOEUF): Gene-level degree of mutational constraints
        AF: (as Allele frequency): Allele frequency in gnomAD database
        5UTR_length: (as 5'UTR length) : The length of 5'UTR
        3UTR_length: (as 3'UTR length) : The length of 3'UTR
        Transcript_length: (as Transcript length) : The length of the transcript
'''

# we expect to match names of columns to your datasets
cols = ['downstream_exon_count', 'last_exon', 'PTC_to_start_codon',
        'dist_to_stop_codon', 'PTC_exon_length', 'PTC_to_intron','upstream_exon_count',
        'mRNA_half_life','50nt_to_last_EJ','LOEUF', 'AF', '5UTR_length', '3UTR_length', 'Transcript_length']

df[cols].head(1)
```

```python
df[cols].isnull().sum()
```

```python
df = df.loc[df['LOEUF'].notnull()].reset_index(drop=True)
df.shape
```

```python
regr = RandomForestRegressor(max_features = 3, n_estimators=10000, n_jobs=-1)
```

```python
# applying cross-valiation method for evaluation
scores = cross_val_score(regr, df[cols], df['NMD_efficiency'], cv=5)
print(f'r2: {np.mean(scores):5f}, std: {np.std(scores):5f}')
```

```python
X = df[cols]
y = df['NMD_efficiency']
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=0)
```

```python
regr.fit(X_train, y_train)
regr.score(X_test, y_test)
print(f'r2: {regr.score(X_test, y_test):5f}')
```

### Feature–target correlation analysis

To complement the model-derived feature importance analysis, Pearson correlation coefficients and corresponding p-values are computed between individual features and NMD_efficiency to assess the strength and statistical significance of linear associations between features and the target variable.

```python
df.columns
```

```python
importance_lookup = dict(zip(importance_df["feature"], importance_df["importance"]))

def format_p_value(p):
    if p < 0.001:
        return "p < 0.001"
    else:
        return f"p = {p:.3f}"

def compute_feature_correlations(df, features, target="NMD_efficiency"):
    results = []
    for feat in features:
        x = df[feat].dropna()
        y = df.loc[x.index, target].dropna()
        if len(x) > 5:  # avoid tiny groups
            pear, p_p = pearsonr(x, y)
            spear, p_s = spearmanr(x, y)            
            results.append({
                "feature": feat,
                "pearson_r": pear,
                # "spearman_r": spear,
                "p_value": format_p_value(p_p),
                "importance": importance_lookup.get(feat, None)
            })
    return pd.DataFrame(results)

features = ["downstream_exon_count", "PTC_to_start_codon", "PTC_exon_length", "3UTR_length", "PTC_to_intron", "last_exon", "LOEUF", "Transcript_length", "mRNA_half_life", "5UTR_length", 
            "upstream_exon_count", "AF", "50nt_to_last_EJ"]

corr_df = compute_feature_correlations(df, features)
corr_df
```

## 2. Training a model using NMD Scanner annotations on the TCGA benchmark

In this section, a regression model is trained using variant-level features generated by the NMD-Scanner and experimentally measured NMD_efficiency values from the original TCGA dataset.

### Integration of NMD-Scanner annotations with TCGA benchmark data

Variant-level annotations generated by the NMD-Scanner are merged with the TCGA benchmark dataset from the NMD-efficiency study. Variants are matched based on transcript identifiers and genomic coordinates after minor preprocessing to ensure consistency between annotation formats. The resulting dataset combines NMD-Scanner-derived features with experimentally measured *NMD_efficiency* values and serves as the input for all subsequent analyses in this section.

```python
# merge NMD_Scanner Output von TCGA with the NMD_efficiency score from the Original TCGA Benchmark
#df = pd.read_csv('~/NMD/nmd-variant-effect-prediction/resources/Output-260825/tcga_dataset_final_nmd_results.csv') # nmd scanner output on tcga
df = pd.read_csv('~/NMD/nmd-variant-effect-prediction/resources/Output-210925/tcga_dataset_final_nmd_results.csv') # nmd scanner output on tcga
df2 = pd.read_csv('~/NMD/nmd-variant-effect-prediction/resources/TCGA_benchmark/tcga_dataset.csv')

# 1. Remove transcript version number from NMD Scanner output (df)
df['transcript_id'] = df['transcript_id'].str.split('.').str[0]

# 2. Adjust start_variant position (+1) to match TCGA coordinates
df['start_variant'] = df['start_variant'] + 1
#df['end_variant'] = df['end_variant'] - 1

# 3. Rename TCGA columns to match NMD Scanner
df2 = df2.rename(columns={
    'Transcript_ID': 'transcript_id',
    'start': 'start_variant',
    'end': 'end_variant'
})

# 4. Select only relevant columns from TCGA
df2_subset = df2[['transcript_id', 'start_variant', 'end_variant', 'NMD_efficiency']]

# 5. Merge
df_merged = pd.merge(df, df2_subset, on=['transcript_id', 'start_variant', 'end_variant'], how='inner')

print("Merged shape:", df_merged.shape)
print("Missing NMD_efficiency values:", df_merged['NMD_efficiency'].isnull().sum())
print(df_merged[['transcript_id', 'start_variant', 'end_variant', 'NMD_efficiency']].head())
```
